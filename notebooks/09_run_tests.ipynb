# Week 2 Day 5 — Run the full test suite
Runs `tests/test_pricing.py` via pytest and shows the results inline.

```python
import subprocess, sys

result = subprocess.run(
    [sys.executable, '-m', 'pytest', 'tests/test_pricing.py', '-v', '--tb=short'],
    capture_output=True,
    text=True,
    cwd='..'
)

print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
```
